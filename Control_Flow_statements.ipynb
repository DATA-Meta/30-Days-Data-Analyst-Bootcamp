## Control Flow statements
Control flow statements are used to determine the order in which instructions are executed in a program. They allow you to make decisions, repeat actions, and control the flow of execution based on certain conditions. The main types of control flow statements include:Conditional statements, loops, and branching statements.

# IF Else Condition

write a program to do display hello if number entered by user is a multiple of 5 means divisible by 5, otherwise display goodbye.

```python
a = int(input("Enter a number: "))  
if a%5==0:
    print("Hello")
else:
    print("Goodbye")
```

# accept the city and print its monument
# Delhi- redfort, Agra- Taj Mahal, Jaipur- Jal Mahal, orissa konak temple, hydrabad charminar, other - record not found

```python
city = input("Enter the city name: ")
if city == "Delhi":
    print("the city monument is Red Fort")
elif city == "Agra":
    print("the city monument is Taj Mahal")
elif city == "Jaipur":
    print("the city monument is Jal Mahal")
elif city == "orissa":
    print("the city monument is Konak Sun Temple")
elif city == "hyderabad":
    print("the city monument is Charminar")
else:
    print("no record found for the city monument")
```
