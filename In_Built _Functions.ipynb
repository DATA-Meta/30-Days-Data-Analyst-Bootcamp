# Builtin Functions in python

# abs() Function

```python
x = -7.25
print(x)
print(abs(x))

x = abs(-7.25)
print(x)
```

# Bin Function

```python
x = bin(13)   # 8 4 2 1
print(x)      # 1 1 0 1

y = bin(36)   # 32 16 8 4 2 1
print(y)      # 1  0  0 1 0 0
```

# Bytes Function

```python
x = bytes(4)   
```

# Python chr() Function

```python
x = chr(97)
print(x)

x = chr(98)
print(x)

x = chr(100)
print(x)

x = chr(92)
print(x)

x = chr(90)
print(x)
```

# complex function

```python
x = complex(3, 5)
print(x)

y = complex(2.5, 3.96)
print(y)

z = complex('2')
print(z)
```

# float() Function

```python
x = float(3)
print(x)

x =  float(3.56)
print(x)
```

# int() Function

```python
x = int(12.56)
print(x)

y = int(846.72)
print(y)
```

# str() Function

```python
a = str(2568)
print(a)
print(type(a))


y = str('2368')
print(y)
print(type(y))
```

# help() Function
