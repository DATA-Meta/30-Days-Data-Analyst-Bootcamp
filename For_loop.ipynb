# For Loop

```python
fruits = ["apple", "banana", "cherry"] # list use []
# here i want to see these fruits print horizontally
for x in fruits:
    print(x)
```

# Looping through a string

```python
for x in "banana":
    print(x)
```

# Range function in for loop

```python
for x in range(25):
    print(x)
```

```python
for x in range(3,25):
    print(x)
```
