# program to calculate electricity bill, 500 unit = 5£ , 700 units = 10£ , print bill dates , fine date , address for payment bill for 1000 people

```python
# 500 units = 5 per unit, 700 units = 10 per unit, 
# 1000 units = 15 per unit
# if more than 1000 units, then 20 per unit

def electricity_bill(n):
    if n<=500:
        print("your bill is in Rs: " , n*5)
    elif n>500 and n<=700:
        print("you bill in Rs:" , n*10)
    elif n>700 and n<=1000:
        print("your bill is Rs: ", n*15)
    elif n>1000:
        print("your bill is in Rs: ", n*20)
```
